```python
# PASO 5 - TEST A/A/B
group_users = logs_filtered.groupby('exp_id')['user_id'].nunique()

print('Usuarios por grupo:')
print(group_users)
```

```python
# Comparar los grupos A/A: 246 vs 247
from statsmodels.stats.proportion import proportions_ztest

def test_proportions(group1, group2, event):
    # Usuarios únicos de cada grupo
    users_group1 = logs_filtered[logs_filtered['exp_id'] == group1]['user_id'].nunique()
    users_group2 = logs_filtered[logs_filtered['exp_id'] == group2]['user_id'].nunique()
    
    # Usuarios que realizaron el evento
    event_users_group1 = logs_filtered[
        (logs_filtered['exp_id'] == group1) &
        (logs_filtered['event_name'] == event)
    ]['user_id'].nunique()
    
    event_users_group2 = logs_filtered[
        (logs_filtered['exp_id'] == group2) &
        (logs_filtered['event_name'] == event)
    ]['user_id'].nunique()
    
    # Prueba de proporciones
    count = [event_users_group1, event_users_group2]
    nobs = [users_group1, users_group2]
    
    stat, pvalue = proportions_ztest(count, nobs)
    
    # Proporciones
    proportion1 = event_users_group1 / users_group1
    proportion2 = event_users_group2 / users_group2
    
    return proportion1, proportion2, pvalue
```

```python
# Evento MainScreenAppear
p1, p2, pvalue = test_proportions(246, 247, 'MainScreenAppear')

print(f'Grupo 246: {p1:.4f} ({p1*100:.2f}%)')
print(f'Grupo 247: {p2:.4f} ({p2*100:.2f}%)')
print(f'Valor p: {pvalue:.4f}')
```

<b>Comentario del Revisor - 1a iteración</b> <a class="tocSkip"></a><br>

<b>Éxito</b> -  Al implementar el test A/A/B usaste <code>proportions_ztest</code> para comparar proporciones y reportaste p-values, por ejemplo 'Valor p: 0.7571'. También mencionaste la corrección de Bonferroni en la conclusión. Esto muestra que aplicaste un enfoque estadístico apropiado para pruebas de proporciones.

```python
# Comparación 246 vs 247 - MainScreenAppear
events = [
    'MainScreenAppear',
    'OffersScreenAppear',
    'CartScreenAppear',
    'PaymentScreenSuccessful'
]

for event in events:
    p1, p2, pvalue = test_proportions(246, 247, event)
    
    print(f'\nEvento: {event}')
    print(f'Grupo 246: {p1:.4f} ({p1*100:.2f}%)')
    print(f'Grupo 247: {p2:.4f} ({p2*100:.2f}%)')
    print(f'Valor p: {pvalue:.4f}')
    
    if pvalue < 0.05:
        print('Resultado: diferencia estadísticamente significativa')
    else:
        print('Resultado: no hay diferencia estadísticamente significativa')
```

```python
for event in events:
    p1, p2, pvalue = test_proportions(247, 248, event)
    
    print(f'\nEvento: {event}')
    print(f'Grupo 247: {p1:.4f} ({p1*100:.2f}%)')
    print(f'Grupo 248: {p2:.4f} ({p2*100:.2f}%)')
    print(f'Valor p: {pvalue:.4f}')
    
    if pvalue < 0.05:
        print('Resultado: diferencia estadísticamente significativa')
    else:
        print('Resultado: no hay diferencia estadísticamente significativa')
```

```python
# Comparación 248 vs controles convinados
def test_combined_control(event):
    # Usuarios únicos de los grupos de control combinados
    control_users = logs_filtered[
        logs_filtered['exp_id'].isin([246, 247])
    ]['user_id'].nunique()
    
    # Usuarios únicos del grupo 248
    test_users = logs_filtered[
        logs_filtered['exp_id'] == 248
    ]['user_id'].nunique()
    
    # Usuarios de control que realizaron el evento
    control_event_users = logs_filtered[
        (logs_filtered['exp_id'].isin([246, 247])) &
        (logs_filtered['event_name'] == event)
    ]['user_id'].nunique()
    
    # Usuarios del grupo 248 que realizaron el evento
    test_event_users = logs_filtered[
        (logs_filtered['exp_id'] == 248) &
        (logs_filtered['event_name'] == event)
    ]['user_id'].nunique()
    
    # Prueba Z de proporciones
    count = [control_event_users, test_event_users]
    nobs = [control_users, test_users]
    
    stat, pvalue = proportions_ztest(count, nobs)
    
    # Proporciones
    control_proportion = control_event_users / control_users
    test_proportion = test_event_users / test_users
    
    return control_proportion, test_proportion, pvalue
```

```python

for event in events:
    p_control, p_test, pvalue = test_combined_control(event)
    
    print(f'\nEvento: {event}')
    print(f'Controles 246+247: {p_control:.4f} ({p_control*100:.2f}%)')
    print(f'Grupo 248: {p_test:.4f} ({p_test*100:.2f}%)')
    print(f'Valor p: {pvalue:.4f}')
    
    if pvalue < 0.05:
        print('Resultado: diferencia estadísticamente significativa')
    else:
        print('Resultado: no hay diferencia estadísticamente significativa')
```
