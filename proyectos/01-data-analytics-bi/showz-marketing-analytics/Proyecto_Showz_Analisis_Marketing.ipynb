```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Cargar los datasets
visits = pd.read_csv('/datasets/visits_log_us.csv')
orders = pd.read_csv('/datasets/orders_log_us.csv')
costs = pd.read_csv('/datasets/costs_us.csv')

# Mostrar las primeras filas de cada dataset
print('VISITS')
display(visits.head())

print('ORDERS')
display(orders.head())

print('COSTS')
display(costs.head())
```

```python
# Convertir nombres a minúsculas y reemplazar espacios por guiones bajos
visits.columns = visits.columns.str.lower().str.replace(' ', '_')
orders.columns = orders.columns.str.lower().str.replace(' ', '_')
costs.columns = costs.columns.str.lower().str.replace(' ', '_')

# validar los cambios
print('Columnas de visits:', visits.columns.tolist())
print('Columnas de orders:', orders.columns.tolist())
print('Columnas de costs:', costs.columns.tolist())
```

```python
# Crear columnas auxiliares para realizar los análisis temporales

visits['session_day'] = visits['start_ts'].dt.date
visits['session_week'] = visits['start_ts'].dt.to_period('W').apply(lambda x: x.start_time)
visits['session_month'] = visits['start_ts'].dt.to_period('M').dt.to_timestamp()

# Calcular la duración de cada sesión en segundos
visits['session_duration'] = (
    visits['end_ts'] - visits['start_ts']
).dt.total_seconds()

# Revisar el resultado
display(visits.head())
```

```python
# Calcular DAU: usuarios únicos por día
dau = visits.groupby('session_day')['uid'].nunique()

print('DAU promedio:', round(dau.mean(), 2))
display(dau.head(10))
```

```python
# Gráfico del DAU a lo largo del tiempo
dau.plot(figsize=(12, 5))

plt.title('Usuarios activos diarios (DAU)')
plt.xlabel('Fecha')
plt.ylabel('Usuarios únicos')
plt.grid(True)
plt.show()
```

```python
# Calcular WAU: usuarios únicos por semana
wau = visits.groupby('session_week')['uid'].nunique()

print('WAU promedio:', round(wau.mean(), 2))
display(wau.head(10))
```

```python
# Gráfico del WAU
wau.plot(figsize=(12, 5))

plt.title('Usuarios activos semanales (WAU)')
plt.xlabel('Semana')
plt.ylabel('Usuarios únicos')
plt.grid(True)
plt.show()
```

```python
# Calcular MAU: usuarios únicos por mes
mau = visits.groupby('session_month')['uid'].nunique()

print('MAU promedio:', round(mau.mean(), 2))
display(mau)
```

```python
# Gráfico del MAU

mau.plot(figsize=(12, 5))

plt.title('Usuarios activos mensuales (MAU)')
plt.xlabel('Mes')
plt.ylabel('Usuarios únicos')
plt.grid(True)
plt.show()
```

```python
# Calcular Sticky Factor utilizando el promedio de DAU y MAU
#El Sticky Factor indica qué tan frecuentemente los usuarios regresan al servicio.
sticky_factor = dau.mean() / mau.mean() * 100

print('Sticky Factor:', round(sticky_factor, 2), '%')
```

```python

# ==========================================
# FRECUENCIA DE RETORNO - STICKY FACTOR
# ==========================================

# DAU: usuarios únicos por día
dau_total = visits.groupby('session_day')['uid'].nunique().mean()

# WAU: usuarios únicos por semana
wau_total = visits.groupby('session_week')['uid'].nunique().mean()

# MAU: usuarios únicos por mes
mau_total = visits.groupby('session_month')['uid'].nunique().mean()

# Mostrar las métricas utilizadas
print('DAU promedio:', round(dau_total, 2))
print('WAU promedio:', round(wau_total, 2))
print('MAU promedio:', round(mau_total, 2))

# Sticky Factor diario/mensual
sticky_factor_daily = dau_total / mau_total

# Sticky Factor semanal/mensual
sticky_factor_weekly = wau_total / mau_total

# Mostrar resultados
print('\nSticky Factor diario/mensual:',
      round(sticky_factor_daily, 4))

print('Sticky Factor semanal/mensual:',
      round(sticky_factor_weekly, 4))

# Mostrar también como porcentaje
print('\nSticky Factor DAU/MAU:',
      round(sticky_factor_daily * 100, 2), '%')

print('Sticky Factor WAU/MAU:',
      round(sticky_factor_weekly * 100, 2), '%')
```

```python

# Primera visita de cada usuario
first_visit = (
    visits.groupby('uid')['start_ts']
    .min()
    .reset_index()
)

first_visit.columns = ['uid', 'first_visit_ts']

display(first_visit.head())
```

```python
# Primera compra de cada usuario
first_order = (
    orders.groupby('uid')['buy_ts']
    .min()
    .reset_index()
)

first_order.columns = ['uid', 'first_order_ts']

display(first_order.head())
```

```python
# Unir la primera visita con la primera compra
conversion = first_visit.merge(
    first_order,
    on='uid',
    how='inner'
)

display(conversion.head())
```

```python
# Calcular tiempo desde la primera visita hasta la primera compra
conversion['conversion_time'] = (
    conversion['first_order_ts'] - conversion['first_visit_ts']
)

# Convertir a días
conversion['conversion_days'] = (
    conversion['conversion_time'].dt.total_seconds() / 86400
)

display(conversion.head())
```

```python
# Estadisticas
print('Tiempo promedio hasta la primera compra:',
      round(conversion['conversion_days'].mean(), 2), 'días')

print('Mediana del tiempo de conversión:',
      round(conversion['conversion_days'].median(), 2), 'días')

print('Percentil 90:',
      round(conversion['conversion_days'].quantile(0.90), 2), 'días')
```

```python
# LTV — Lifetime Value
# El LTV indica cuánto ingreso genera, en promedio, un cliente durante un período determinado.

# Obtener el primer pedido de cada cliente
first_order = (
    orders.groupby('uid')['buy_ts']
    .min()
    .reset_index()
)

first_order.columns = ['uid', 'first_order_ts']

# Crear el mes de la primera compra
first_order['cohort_month'] = (
    first_order['first_order_ts']
    .dt.to_period('M')
    .dt.to_timestamp()
)

display(first_order.head())
```

```python
# Añadir la cohorte a cada pedido
orders_cohort = orders.merge(
    first_order[['uid', 'cohort_month']],
    on='uid',
    how='left'
)

# Crear el mes de cada pedido
orders_cohort['order_month'] = (
    orders_cohort['buy_ts']
    .dt.to_period('M')
    .dt.to_timestamp()
)

display(orders_cohort.head())
```

```python
# Calcular los ingresos por cohorte y mes
cohort_revenue = (
    orders_cohort
    .groupby(['cohort_month', 'order_month'])['revenue']
    .sum()
    .unstack()
    .fillna(0)
)

display(cohort_revenue)
```

```python
# Ingresos acumulados por cohorte
cumulative_revenue = cohort_revenue.cumsum(axis=1)

display(cumulative_revenue)
```

```python
# Numero de Clientes por Cohorte

cohort_users = (
    first_order
    .groupby('cohort_month')['uid']
    .nunique()
)

display(cohort_users)
```

<b>Comentario del revisor</b> <a class="tocSkip"></a><br>

<b>Atención</b> - La construcción de las cohortes está bien realizada: identificaste el mes de la primera compra, calculaste los ingresos por cohorte y obtuviste correctamente el número de clientes de cada una.



</div>


```python
# Calcular LTV acumulado por cohorte

ltv = cumulative_revenue.div(cohort_users, axis=0)

display(ltv.round(2))
```

```python
# LTV acumulado después de 6 meses
m6_cum_ltv = ltv.iloc[:, :6].iloc[:, -1].mean()

print('LTV acumulado promedio a 6 meses:', round(m6_cum_ltv, 2))
```

La evolución mensual permite identificar períodos en los que los clientes realizan compras de mayor o menor valor.

```python
# Gasto de Marketing a largo tiempo

# Crear columna con el mes
costs['cost_month'] = (
    costs['dt']
    .dt.to_period('M')
    .dt.to_timestamp()
)

# Gasto mensual
costs_by_month = (
    costs.groupby('cost_month')['costs']
    .sum()
)

display(costs_by_month)

# Grafico
costs_by_month.plot(
    figsize=(12, 5)
)

plt.title('Gasto de marketing a lo largo del tiempo')
plt.xlabel('Mes')
plt.ylabel('Gasto')
plt.grid(True)
plt.show()
```

```python
# Gasto mensual por fuente
costs_source_month = (
    costs.groupby(['cost_month', 'source_id'])['costs']
    .sum()
    .unstack(fill_value=0)
)

display(costs_source_month.head())

#Grafico
costs_source_month.plot(
    figsize=(14, 6)
)

plt.title('Gasto mensual por fuente de anuncios')
plt.xlabel('Mes')
plt.ylabel('Gasto')
plt.grid(True)
plt.show()
```

```python
# CAC — Costo de adquisición de clientes
# CAC = Gasto de marketing / Número de clientes adquiridos

# Obtener la primera visita de cada usuario
first_visit_source = (
    visits.sort_values('start_ts')
    .drop_duplicates('uid')[['uid', 'source_id']]
)

display(first_visit_source.head())
```

```python
# Obtener los usuarios que realizaron al menos una compra
customers = orders[['uid']].drop_duplicates()

# Añadir la fuente de adquisición
customers_source = customers.merge(
    first_visit_source,
    on='uid',
    how='left'
)

display(customers_source.head())
```

```python
# Clientes adquiridos por fuente
customers_by_source = (
    customers_source.groupby('source_id')['uid']
    .nunique()
)

display(customers_by_source)
```

```python
# Calcular CAC por fuente

# Combinar gasto y número de clientes
cac_by_source = pd.DataFrame({
    'marketing_cost': costs_by_source,
    'customers': customers_by_source
})

# Calcular CAC
cac_by_source['cac'] = (
    cac_by_source['marketing_cost'] /
    cac_by_source['customers']
)

display(cac_by_source.sort_values('cac'))

# Grafico
cac_by_source['cac'].sort_values().plot(
    kind='bar',
    figsize=(12, 5)
)

plt.title('CAC por fuente de anuncios')
plt.xlabel('Fuente de anuncios')
plt.ylabel('CAC')
plt.xticks(rotation=0)
plt.grid(axis='y')
plt.show()
```

 Interpretación del CAC

El CAC indica cuánto dinero debe invertir Showz en marketing para adquirir un cliente.

Un CAC bajo es favorable porque significa que la empresa puede adquirir clientes gastando menos dinero.

Sin embargo, el CAC por sí solo no permite determinar qué fuente es más rentable. Una fuente puede tener un CAC relativamente alto pero generar clientes con un LTV mucho mayor.

Por esta razón, compararemos el CAC con los ingresos y el LTV mediante el ROMI.

```python
# ROMI — Retorno de la inversión en marketing
# ROMI = (Ingresos generados por marketing - Gasto de marketing)/ Gasto de marketing

# Añadir la fuente de adquisición a cada pedido
orders_source = orders.merge(
    first_visit_source,
    on='uid',
    how='left'
)

display(orders_source.head())
```

```python
# Ingresos generados por cada fuente
revenue_by_source = (
    orders_source.groupby('source_id')['revenue']
    .sum()
    .sort_values(ascending=False)
)

display(revenue_by_source)
```

```python
# Calcular ROMI
# Crear tabla para calcular ROMI
romi_by_source = pd.DataFrame({
    'marketing_cost': costs_by_source,
    'revenue': revenue_by_source
})

# Calcular ROMI
romi_by_source['romi'] = (
    (romi_by_source['revenue'] - romi_by_source['marketing_cost'])
    / romi_by_source['marketing_cost']
)

display(romi_by_source.sort_values('romi', ascending=False))
```

```python
# ROMI expresado en porcentaje
romi_by_source['romi_pct'] = romi_by_source['romi'] * 100

display(
    romi_by_source
    .sort_values('romi_pct', ascending=False)
)
```

```python
# Gráfico del ROMI
romi_by_source['romi_pct'].sort_values().plot(
    kind='bar',
    figsize=(12, 5)
)

plt.title('ROMI por fuente de anuncios')
plt.xlabel('Fuente de anuncios')
plt.ylabel('ROMI (%)')
plt.xticks(rotation=0)
plt.axhline(0, linewidth=1)
plt.grid(axis='y')
plt.show()
```

```python
# Crear tabla resumen de las métricas de marketing

marketing_summary = pd.DataFrame({
    'marketing_cost': costs_by_source,
    'customers': customers_by_source,
    'cac': cac_by_source['cac'],
    'revenue': revenue_by_source,
    'romi_pct': romi_by_source['romi_pct']
})

# Ordenar por ROMI de mayor a menor
marketing_summary = marketing_summary.sort_values(
    'romi_pct',
    ascending=False
)

display(marketing_summary.round(2))
```

```python

# Fuente con ROMI más alto y más bajo

source_highest_romi = marketing_summary['romi_pct'].idxmax()
highest_romi = marketing_summary['romi_pct'].max()

source_lowest_romi = marketing_summary['romi_pct'].idxmin()
lowest_romi = marketing_summary['romi_pct'].min()

print('Fuente con ROMI más alto:', source_highest_romi)
print('ROMI más alto:', round(highest_romi, 2), '%')

print('\nFuente con ROMI más bajo:', source_lowest_romi)
print('ROMI más bajo:', round(lowest_romi, 2), '%')
```

```python
# Fuente con CAC más bajo y más alto

source_lowest_cac = marketing_summary['cac'].idxmin()
lowest_cac = marketing_summary['cac'].min()

source_highest_cac = marketing_summary['cac'].idxmax()
highest_cac = marketing_summary['cac'].max()

print('Fuente con CAC más bajo:', source_lowest_cac)
print('CAC más bajo:', round(lowest_cac, 2))

print('\nFuente con CAC más alto:', source_highest_cac)
print('CAC más alto:', round(highest_cac, 2))
```
